```python
from dataclasses import dataclass
import numpy as np
import pandas as pd
import mesa
```

```python
@dataclass
class Vintage:
    """Représente une génération de capital fixe (un équipement) détenue par une firme."""
    uc: float           # Coût unitaire d'exploitation (hors amortissement)
    kappa: float        # Coût unitaire en capital
    k_init: float       # Capacité productive initiale acquise (Delta k_v)
    k: float            # Capacité productive survivante actuelle k_v(t)
    birth_period: int   # Période d'installation tau_v
```

```python
class FirmAgent(mesa.Agent):
    """Représente une firme qui gère un portefeuille de capitaux (vintages) et fixe ses prix."""
    def __init__(
        self,
        model: mesa.Model,
        rule: str,
        initial_vintages: list,
        uc_b: float,
        kappa_b: float,
        price: float,
        target_share: float,
        budget: float,
        u_init: float = 0.85,
        entry_period: int = 0
    ):
        super().__init__(model)
        self.rule = rule
        self.vintages = initial_vintages
        self.uc_b = float(uc_b)
        self.kappa_b = float(kappa_b)
        self.uc_a = None
        self.kappa_a = None
        
        self.price = float(price)
        self.target_share = float(target_share)
        self.realized_share = float(target_share)
        self.budget = float(budget)
        self.u = float(u_init)
        self.entry_period = int(entry_period)
        self.sales = 0.0
        self.profit = 0.0
        self.k_surv = self.total_capacity

    @property
    def total_capacity(self) -> float:
        return sum(v.k for v in self.vintages)

    @property
    def mean_uc(self) -> float:
        K = self.total_capacity
        if K <= 0:
            return self.uc_b
        return sum(v.uc * v.k for v in self.vintages) / K

    @property
    def mean_kappa(self) -> float:
        K = self.total_capacity
        if K <= 0:
            return self.kappa_b
        return sum(v.kappa * v.k for v in self.vintages) / K

    def normal_profit_rate(self) -> float:
        K = self.total_capacity
        if K <= 0 or self.mean_kappa <= 0:
            return 0.0
        return (self.price - self.mean_uc) / self.mean_kappa

    def set_price(self, p_leader: float, is_shortage: bool):
        p_floor = self.mean_uc
        if not is_shortage:
            eps_i = max(0.0, self.model.gamma * (self.model.u_n - self.u) + self.model.eps_0)
            eps_i = min(eps_i, 1.0 - 1e-12)
            self.price = max(p_floor, p_leader * (1.0 - eps_i))
        else:
            z_i = self.model.draw_truncated_normal()
            self.price = max(p_floor, self.price, p_leader * (1.0 + z_i))
        assert self.price >= p_floor - 1e-12, f"Violation de p_i(t) >= uc_bar_i(t): {self.price} < {p_floor}"

    def profitable_supply(self) -> float:
        return sum(v.k for v in self.vintages if v.uc <= self.price)

    def produce_and_account(self):
        sorted_vintages = sorted(self.vintages, key=lambda v: (v.uc, -v.birth_period))
        remaining_output = self.sales
        net_profit = 0.0
        for v in sorted_vintages:
            if remaining_output <= 0.0:
                break
            q_v = min(v.k, remaining_output)
            net_profit += (self.price - v.uc) * q_v
            remaining_output -= q_v
        self.profit = net_profit
        self.budget += self.profit

    def scrap_vintages(self, current_t: int):
        surviving = []
        for v in self.vintages:
            age = current_t - v.birth_period + 1
            if age < self.model.T and v.uc <= self.price:
                v.k = v.k_init
                surviving.append(v)
        self.vintages = surviving
        self.k_surv = self.total_capacity

    def _frontier_capacity_share(self) -> float:
        """Calcule la part de capacité incarnée dans la technique frontière (uc_a, kappa_a)."""
        if self.uc_a is None or self.k_surv <= 0.0:
            return 0.0
        k_frontier = sum(
            v.k for v in self.vintages
            if np.isclose(v.uc, self.uc_a) and np.isclose(v.kappa, self.kappa_a)
        )
        return k_frontier / self.k_surv

    def innovate_and_invest(self, current_t: int, forced_draw: tuple = None):
        """Étape 5 : Innovation DLF, adoption (Shaikh vs Okishio) et investissement."""
        # --- 1. Vérification de la consolidation pour une firme Shaikh ---
        can_search = True
        if self.rule == "shaikh" and self.uc_a is not None:
            if self._frontier_capacity_share() >= self.model.gamma_thresh:
                # Promotion de la technique frontière en nouvelle référence
                self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
                self.uc_a, self.kappa_a = None, None
            else:
                # Consolidation en cours : R&D suspendue
                can_search = False

        # --- 2. Tirage et évaluation d'une innovation candidate ---
        if can_search:
            d_uc, d_kappa = forced_draw if forced_draw is not None else self.model.draw_innovation()
            uc_plus = self.uc_b / (1.0 + d_uc)
            kappa_plus = self.kappa_b / (1.0 + d_kappa)

            if self.rule == "okishio":
                r_b = (self.price - self.uc_b) / self.kappa_b
                r_plus = (self.price - uc_plus) / kappa_plus
                if r_plus > r_b:
                    self.uc_b, self.kappa_b = uc_plus, kappa_plus
                    self.uc_a, self.kappa_a = None, None
            elif self.rule == "shaikh":
                if uc_plus < self.uc_b:  # Critère de réduction des coûts (Delta uc > 0)
                    self.uc_a, self.kappa_a = uc_plus, kappa_plus

        # --- 3. Allocation du budget d'investissement B_i(t) ---
        assert self.budget >= -1e-12, f"Budget négatif avant investissement : {self.budget}"
        if self.budget <= 0.0:
            return

        min_kappa_avail = min(self.kappa_b, self.kappa_a) if self.kappa_a is not None else self.kappa_b
        if self.k_surv <= 1e-12 and self.budget < min_kappa_avail:
            return

        budget_before = self.budget
        new_birth = current_t + 1

        if self.rule == "okishio":
            # Okishio investit tout son budget dans la technique offrant le taux de profit maximal
            dk_b = budget_before / self.kappa_b
            cost_total = self.kappa_b * dk_b
            assert cost_total <= budget_before + 1e-9, f"Dépassement de budget B_i(t): {cost_total} > {budget_before}"
            if dk_b > 0.0:
                self.vintages.append(Vintage(self.uc_b, self.kappa_b, dk_b, dk_b, new_birth))
                self.budget = max(0.0, budget_before - cost_total)

        elif self.rule == "shaikh":
            if self.uc_a is None:
                # Pas de technique frontière active : investit tout le budget dans (uc_b, kappa_b)
                dk_b = budget_before / self.kappa_b
                cost_total = self.kappa_b * dk_b
                assert cost_total <= budget_before + 1e-9, f"Dépassement de budget B_i(t): {cost_total} > {budget_before}"
                if dk_b > 0.0:
                    self.vintages.append(Vintage(self.uc_b, self.kappa_b, dk_b, dk_b, new_birth))
                    self.budget = max(0.0, budget_before - cost_total)
            else:
                # Résolution exacte du portefeuille (dk_b, dk_a) minimisant uc_bar(t+1)
                k_min = self.sales / self.model.u_n
                I_req = max(0.0, k_min - self.k_surv)
                min_kappa = min(self.kappa_b, self.kappa_a)
                target_exp = min(I_req, budget_before / min_kappa)

                # On veut maximiser dk_a (car uc_a < uc_b) tout en assurant dk_a + dk_b >= target_exp
                if self.kappa_a <= self.kappa_b:
                    # La technique frontière est aussi la moins chère en capital : 100% du budget sur a !
                    dk_a = budget_before / self.kappa_a
                    dk_b = 0.0
                else:
                    # La technique frontière est plus intensive en capital (kappa_a > kappa_b)
                    max_dk_a_budget = budget_before / self.kappa_a
                    if max_dk_a_budget >= target_exp:
                        # Même en n'achetant que du a, on couvre largement I_req -> 100% sur a !
                        dk_a = max_dk_a_budget
                        dk_b = 0.0
                    else:
                        # Panachage optimal pour atteindre target_exp sans dépasser le budget B_i(t)
                        dk_a = max(0.0, (budget_before - self.kappa_b * target_exp) / (self.kappa_a - self.kappa_b))
                        dk_b = max(0.0, target_exp - dk_a)

                cost_total = self.kappa_b * dk_b + self.kappa_a * dk_a
                assert cost_total <= budget_before + 1e-9, f"Dépassement de budget B_i(t): {cost_total} > {budget_before}"
                if dk_a > 0.0:
                    self.vintages.append(Vintage(self.uc_a, self.kappa_a, dk_a, dk_a, new_birth))
                if dk_b > 0.0:
                    self.vintages.append(Vintage(self.uc_b, self.kappa_b, dk_b, dk_b, new_birth))
                self.budget = max(0.0, budget_before - cost_total)

                # Vérification immédiate de consolidation après installation du nouveau capital
                self.k_surv = self.total_capacity
                if self._frontier_capacity_share() >= self.model.gamma_thresh:
                    self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
                    self.uc_a, self.kappa_a = None, None
        self.k_surv = self.total_capacity
        assert self.budget >= -1e-9, f"Budget négatif après investissement : {self.budget}"


class RealCompetitionModel(mesa.Model):
    """Modèle complet de concurrence réelle d'Anwar Shaikh (Étapes 1 à 6)."""
    def __init__(
        self,
        D: float = 85.0,
        u_n: float = 0.85,
        T: int = 10,
        gamma: float = 0.03,
        eps_0: float = 0.005,
        mu_z: float = 0.01,
        sigma_z: float = 0.005,
        lam: float = 0.8,
        R: float = 0.05,
        innovation_regime: str = "neutral",
        marx_a: float = 0.08,
        marx_theta: float = -np.pi / 4,
        gamma_thresh: float = 0.75,
        r_star: float = 0.12,
        psi: float = 5.0,
        f_0: float = 0.02,
        K_0: float = 10.0,
        B_0: float = 15.0,
        s_min: float = 0.01,
        tau_grace: int = 20,
        seed: int = 42
    ):
        super().__init__(rng=seed)
        self.D = float(D)
        self.u_n = float(u_n)
        self.T = int(T)
        self.gamma = float(gamma)
        self.eps_0 = float(eps_0)
        self.mu_z = float(mu_z)
        self.sigma_z = float(sigma_z)
        self.lam = float(lam)
        self.R = float(R)
        self.innovation_regime = innovation_regime
        self.marx_a = float(marx_a)
        self.marx_b = (self.R ** 2) / self.marx_a
        self.marx_theta = float(marx_theta)
        self.gamma_thresh = float(gamma_thresh)
        
        # Paramètres d'entrée / sortie (Étape 6)
        self.r_star = float(r_star)
        self.psi = float(psi)
        self.f_0 = float(f_0)
        self.K_0 = float(K_0)
        self.B_0 = float(B_0)
        self.s_min = float(s_min)
        self.tau_grace = int(tau_grace)
        
        self.rng = np.random.default_rng(seed)
        self.current_period = 0
        self.r_branch = 0.0

    def draw_truncated_normal(self) -> float:
        while True:
            z = float(self.rng.normal(self.mu_z, self.sigma_z))
            if z >= 0.0:
                return z

    def draw_innovation(self) -> tuple:
        if self.innovation_regime == "neutral":
            angle = self.rng.uniform(0.0, 2.0 * np.pi)
            radius = self.R * np.sqrt(self.rng.uniform(0.0, 1.0))
            return radius * np.cos(angle), radius * np.sin(angle)
        else:
            c, s = np.cos(self.marx_theta), np.sin(self.marx_theta)
            bound = max(self.marx_a, self.marx_b)
            while True:
                x = self.rng.uniform(-bound, bound)
                y = self.rng.uniform(-bound, bound)
                u_rot = x * c + y * s
                v_rot = -x * s + y * c
                if (u_rot / self.marx_a) ** 2 + (v_rot / self.marx_b) ** 2 <= 1.0:
                    return x, y

    def step_1_pricing(self):
        firms = list(self.agents)
        if not firms:
            return
        S_t = sum(f.total_capacity for f in firms)
        p_leader = min(f.price for f in firms)
        is_shortage = S_t < self.D
        for f in firms:
            f.set_price(p_leader=p_leader, is_shortage=is_shortage)
            assert f.price >= f.mean_uc - 1e-12, f"Assertion p_i(t) >= uc_bar_i(t) échouée pour la firme {f.unique_id}"

    def step_2_demand_allocation(self):
        firms = list(self.agents)
        if not firms:
            return
        leader = min(firms, key=lambda f: f.price)
        p_l = leader.price
        sum_rivals_f = 0.0
        for f in firms:
            if f is not leader:
                price_gap = (p_l - f.price) / p_l
                f.target_share = f.target_share * max(0.0, 1.0 + self.lam * price_gap)
                sum_rivals_f += f.target_share
        leader.target_share = max(0.0, 1.0 - sum_rivals_f)
        assert np.isclose(sum(f.target_share for f in firms), 1.0, atol=1e-9), "Assertion sum(f_i(t)) = 1 échouée à l'Étape 2"
        
        q_max = {f: f.profitable_supply() for f in firms}
        d_init = {f: f.target_share * self.D for f in firms}
        q_init = {f: min(q_max[f], d_init[f]) for f in firms}
        
        delta_D = sum(d_init[f] - q_init[f] for f in firms)
        spare = {f: max(0.0, q_max[f] - q_init[f]) for f in firms}
        total_spare = sum(spare.values())
        
        for f in firms:
            if total_spare > 0.0 and delta_D > 0.0:
                f.sales = q_init[f] + spare[f] * min(1.0, delta_D / total_spare)
            else:
                f.sales = q_init[f]
                
        total_sales = sum(f.sales for f in firms)
        for f in firms:
            f.realized_share = (f.sales / total_sales) if total_sales > 0.0 else 0.0
            K_i = f.total_capacity
            f.u = (f.sales / K_i) if K_i > 0.0 else 0.0

    def step_3_production_and_accounting(self):
        for f in self.agents:
            f.produce_and_account()

    def step_4_scrapping(self):
        for f in self.agents:
            f.scrap_vintages(self.current_period)

    def step_5_innovation_and_investment(self):
        for f in self.agents:
            f.innovate_and_invest(self.current_period)

    def _best_practice_technique(self, rule: str, p_ref: float, firms: list) -> tuple:
        """Identifie la technique best-practice de la catégorie 'rule' au prix p_ref."""
        candidates = []
        for f in firms:
            if f.rule == rule:
                candidates.append((f.uc_b, f.kappa_b))
                if f.uc_a is not None:
                    candidates.append((f.uc_a, f.kappa_a))
        if not candidates:
            # Repli si aucune firme de cette catégorie n'est active
            for f in firms:
                candidates.append((f.uc_b, f.kappa_b))
        # Meilleure technique selon le taux de profit au prix leader courant p_ref
        return max(candidates, key=lambda tech: (p_ref - tech[0]) / tech[1])

    def step_6_entry_and_exit(self, capital_value_t: float):
        """Étape 6 : Calcul de r_branch, sorties (exit) puis entrées (entry) avec conservation de somme(f_i)=1."""
        firms = list(self.agents)
        if not firms:
            return

        # 1. Taux de profit agrégé de la branche en période t
        total_profits = sum(f.profit for f in firms)
        self.r_branch = (total_profits / capital_value_t) if capital_value_t > 0.0 else 0.0
        p_leader_now = min(f.price for f in firms)

        # On mémorise les techniques best-practice avant d'éventuelles sorties
        best_shaikh = self._best_practice_technique("shaikh", p_leader_now, firms)
        best_okishio = self._best_practice_technique("okishio", p_leader_now, firms)

        # 2. Sorties (Capital Exit) : marginalisation post-grâce OU épuisement physique & financier
        survivors = []
        for f in firms:
            firm_age = self.current_period - f.entry_period
            marginalized = (firm_age >= self.tau_grace) and (f.realized_share < self.s_min)
            min_kappa_avail = min(f.kappa_b, f.kappa_a) if f.kappa_a is not None else f.kappa_b
            exhausted = (f.total_capacity <= 1e-12) and (f.budget < min_kappa_avail)
            
            if marginalized or exhausted:
                f.remove()  # Retire l'agent du modèle Mesa
            else:
                survivors.append(f)

        # Redistribution proportionnelle des parts cibles f_i aux survivants
        sum_f_surv = sum(f.target_share for f in survivors)
        if sum_f_surv > 0.0:
            for f in survivors:
                f.target_share /= sum_f_surv
        elif survivors:
            for f in survivors:
                f.target_share = 1.0 / len(survivors)

        # 3. Entrées (Capital Entry) via loi de Poisson
        excess_r = max(0.0, self.r_branch - self.r_star)
        n_entry = int(self.rng.poisson(self.psi * excess_r))
        # Sécurité N_entry * f_0 < 1
        if self.f_0 > 0:
            max_entries = int(0.99 / self.f_0)
            n_entry = min(n_entry, max_entries)

        if n_entry > 0:
            # Rééchelonnement des parts des firmes en place par (1 - N_entry * f_0)
            scale_factor = 1.0 - n_entry * self.f_0
            for f in survivors:
                f.target_share *= scale_factor

            next_t = self.current_period + 1
            for _ in range(n_entry):
                rule = "shaikh" if self.rng.random() < 0.5 else "okishio"
                uc_bp, kappa_bp = best_shaikh if rule == "shaikh" else best_okishio
                v_init = Vintage(uc=uc_bp, kappa=kappa_bp, k_init=self.K_0, k=self.K_0, birth_period=next_t)
                FirmAgent(
                    model=self,
                    rule=rule,
                    initial_vintages=[v_init],
                    uc_b=uc_bp,
                    kappa_b=kappa_bp,
                    price=p_leader_now,
                    target_share=self.f_0,
                    budget=self.B_0,
                    u_init=self.u_n,
                    entry_period=next_t
                )

        active_firms = list(self.agents)
        if active_firms:
            assert np.isclose(sum(f.target_share for f in active_firms), 1.0, atol=1e-9), "Assertion sum(f_i(t)) = 1 échouée à l'Étape 6"

    def step(self):
        """Exécute une période complète t -> t+1 dans l'ordre strict des 6 étapes."""
        self.current_period += 1
        
        # Mémorise la valeur du capital installé en période t pour r_branch
        capital_value_t = sum(v.kappa * v.k for f in self.agents for v in f.vintages)
        
        self.step_1_pricing()
        self.step_2_demand_allocation()
        self.step_3_production_and_accounting()
        self.step_4_scrapping()
        self.step_5_innovation_and_investment()
        self.step_6_entry_and_exit(capital_value_t)
```

```python
# --- TEST SPÉCIFIQUE : Décisions d'investissement & Consolidation d'une firme Shaikh ---
m_inv = RealCompetitionModel(u_n=0.85, gamma_thresh=0.75)

# Paramètres communs :
# Technique de référence (b) : uc_b = 0.50, kappa_b = 1.0
# Technique frontière adoptée (a) : uc_a = 0.40, kappa_a = 1.5 (coût d'exploitation plus bas, mais plus chère en capital !)
# Capacité survivante initiale : k_surv = 20.0 (en technique b)
# Ventes q = 85.0 -> Capacité minimale requise K_min = 85.0 / 0.85 = 100.0
# Besoin d'expansion physique : I_req = 100.0 - 20.0 = 80.0 unités de nouvelles machines !

# ==============================================================================
# CAS 1 : Budget serré (B = 100.0) -> PANACHAGE OPTIMAL (dk_a > 0 et dk_b > 0)
# ==============================================================================
# Avec B = 100 :
# - En n'achetant que du a (kappa_a = 1.5), elle n'aurait que 100 / 1.5 = 66.67 machines (< 80.0 requises).
# - En n'achetant que du b (kappa_b = 1.0), 80 machines coûtent 80.0 (il reste 20.0 de surplus).
# - L'algo doit donc convertir 20.0 / (1.5 - 1.0) = 40.0 machines en technique a, et acheter 40.0 en technique b !
f_panache = FirmAgent(m_inv, "shaikh", [Vintage(0.50, 1.0, 20.0, 20.0, 0)], 0.50, 1.0, price=0.70, target_share=1.0, budget=100.0)
f_panache.uc_a, f_panache.kappa_a = 0.40, 1.5  # Déjà adoptée, en cours de consolidation
f_panache.sales = 85.0
f_panache.k_surv = 20.0

f_panache.innovate_and_invest(current_t=1)

# Vérifications Cas 1 :
k_a_panache = sum(v.k for v in f_panache.vintages if np.isclose(v.uc, 0.40))
k_b_new_panache = sum(v.k for v in f_panache.vintages if np.isclose(v.uc, 0.50) and v.birth_period == 2)

assert np.isclose(k_a_panache, 40.0), f"Attendu dk_a = 40.0, obtenu {k_a_panache}"
assert np.isclose(k_b_new_panache, 40.0), f"Attendu dk_b = 40.0, obtenu {k_b_new_panache}"
assert np.isclose(f_panache.total_capacity, 100.0)  # Objectif K_min = 100 atteint !
assert np.isclose(f_panache.budget, 0.0)           # Tout le budget B = 100 est investi !
assert f_panache.uc_a == 0.40                      # Part de a = 40 / 100 = 40% < 75% -> toujours en consolidation !

print(f"[Cas 1 - Panachage] dk_a = {k_a_panache:.1f}, dk_b = {k_b_new_panache:.1f} | Capacité = {f_panache.total_capacity:.1f} | En consolidation ? {f_panache.uc_a is not None}")

# ==============================================================================
# CAS 2 : Budget confortable (B = 135.0) -> 100% FRONTIÈRE + CONSOLIDATION !
# ==============================================================================
# Avec B = 135 :
# - Même en n'achetant que du a (kappa_a = 1.5), elle achète 135 / 1.5 = 90.0 machines (>= 80.0 requises).
# - L'algo doit mettre 100% du budget sur a (dk_a = 90.0, dk_b = 0.0).
# - Capacité totale = 20 (b) + 90 (a) = 110.0 -> Part de a = 90 / 110 = 81.8% >= 75% (gamma_thresh) !
# - La technique frontière a (0.40, 1.5) doit être PROMUE nouvelle référence (uc_b, kappa_b) !
f_riche = FirmAgent(m_inv, "shaikh", [Vintage(0.50, 1.0, 20.0, 20.0, 0)], 0.50, 1.0, price=0.70, target_share=1.0, budget=135.0)
f_riche.uc_a, f_riche.kappa_a = 0.40, 1.5
f_riche.sales = 85.0
f_riche.k_surv = 20.0

f_riche.innovate_and_invest(current_t=1)

k_a_riche = sum(v.k for v in f_riche.vintages if np.isclose(v.uc, 0.40))
assert np.isclose(k_a_riche, 90.0)
assert np.isclose(f_riche.total_capacity, 110.0)
assert np.isclose(f_riche.uc_b, 0.40)    # Promue nouvelle technique de référence !
assert np.isclose(f_riche.kappa_b, 1.5)
assert f_riche.uc_a is None              # Consolidation terminée, R&D prête à repartir !

print(f"[Cas 2 - Consolidation] dk_a = {k_a_riche:.1f}, dk_b = 0.0 | Nouvelle réf uc_b = {f_riche.uc_b:.2f} | R&D réouverte ? {f_riche.uc_a is None}")
print("✅ Algorithme d'investissement et de consolidation Shaikh 100% validé !")
```

```python
import matplotlib.pyplot as plt

def run_simulation(
    n_steps: int = 150,
    innovation_regime: str = "neutral",
    n_firms_init: int = 10,
    D: float = 85.0,
    r_star: float = 0.12,
    seed: int = 42
) -> pd.DataFrame:
    """Initialise et exécute une simulation complète du modèle de concurrence réelle."""
    model = RealCompetitionModel(
        D=D,
        u_n=0.85,
        T=10,
        gamma=0.03,
        eps_0=0.005,
        mu_z=0.01,
        sigma_z=0.005,
        lam=0.8,
        R=0.05,
        innovation_regime=innovation_regime,
        marx_a=0.08,
        marx_theta=-np.pi / 4,
        gamma_thresh=0.75,
        r_star=r_star,
        psi=5.0,
        f_0=0.02,
        K_0=10.0,
        B_0=15.0,
        s_min=0.01,
        tau_grace=20,
        seed=seed
    )

    # Initialisation symétrique à t = 0 : moitié Shaikh, moitié Okishio
    uc_0, kappa_0 = 0.50, 1.0
    p_0 = uc_0 + r_star * kappa_0  # Prix normal initial cohérent avec Shaikh
    k_init_firm = (D / 0.85) / n_firms_init  # Capacité initiale calée sur u_n = 0.85

    for idx in range(n_firms_init):
        rule = "shaikh" if idx % 2 == 0 else "okishio"
        # On étale les âges initiaux des machines entre -9 et 0 pour éviter un renouvellement synchronisé
        birth_0 = -(idx % 10)
        v0 = Vintage(uc=uc_0, kappa=kappa_0, k_init=k_init_firm, k=k_init_firm, birth_period=birth_0)
        FirmAgent(
            model=model,
            rule=rule,
            initial_vintages=[v0],
            uc_b=uc_0,
            kappa_b=kappa_0,
            price=p_0,
            target_share=1.0 / n_firms_init,
            budget=15.0,
            u_init=0.85,
            entry_period=0
        )

    history = []
    for t in range(1, n_steps + 1):
        model.step()
        firms = list(model.agents)
        if not firms:
            break

        total_sales = sum(f.sales for f in firms)
        p_leader = min(f.price for f in firms)
        p_avg = sum(f.price * f.sales for f in firms) / total_sales if total_sales > 0 else p_leader

        shaikh_firms = [f for f in firms if f.rule == "shaikh"]
        okishio_firms = [f for f in firms if f.rule == "okishio"]

        share_shaikh = sum(f.realized_share for f in shaikh_firms)
        share_okishio = sum(f.realized_share for f in okishio_firms)

        kappa_shaikh = np.mean([f.mean_kappa for f in shaikh_firms]) if shaikh_firms else np.nan
        kappa_okishio = np.mean([f.mean_kappa for f in okishio_firms]) if okishio_firms else np.nan

        history.append({
            "t": t,
            "n_firms": len(firms),
            "p_leader": p_leader,
            "p_avg": p_avg,
            "r_branch": model.r_branch,
            "share_shaikh": share_shaikh,
            "share_okishio": share_okishio,
            "kappa_shaikh": kappa_shaikh,
            "kappa_okishio": kappa_okishio,
        })

    return pd.DataFrame(history)


# --- Lancement des deux régimes sur 150 périodes ---
df_neutral = run_simulation(n_steps=150, innovation_regime="neutral", seed=42)
df_marx = run_simulation(n_steps=150, innovation_regime="marx", seed=42)

# --- Visualisation comparative en 4 graphiques ---
fig, axes = plt.subplots(2, 2, figsize=(14, 9), sharex=True)

# 1. Trajectoire des prix (leader vs moyenne)
axes[0, 0].plot(df_neutral["t"], df_neutral["p_leader"], label="Leader (Neutre)", color="tab:blue")
axes[0, 0].plot(df_marx["t"], df_marx["p_leader"], label="Leader (Marx-biaisé)", color="tab:red")
axes[0, 0].set_title("Évolution du prix leader p_leader(t)")
axes[0, 0].set_ylabel("Prix")
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# 2. Taux de profit de la branche vs taux général r*
axes[0, 1].plot(df_neutral["t"], df_neutral["r_branch"], label="r_branch (Neutre)", color="tab:blue")
axes[0, 1].plot(df_marx["t"], df_marx["r_branch"], label="r_branch (Marx-biaisé)", color="tab:red")
axes[0, 1].axhline(0.12, color="black", linestyle="--", label="Taux général r* = 12%")
axes[0, 1].set_title("Gravitation du taux de profit de la branche")
axes[0, 1].set_ylabel("Taux de profit")
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# 3. Parts de marché : Shaikh vs Okishio (Régime biaisé à la Marx)
axes[1, 0].plot(df_marx["t"], df_marx["share_shaikh"], label="Part de marché Shaikh", color="tab:red", linewidth=2)
axes[1, 0].plot(df_marx["t"], df_marx["share_okishio"], label="Part de marché Okishio", color="tab:gray", linestyle="--")
axes[1, 0].set_title("Sélection concurrentielle sous innovation biaisée à la Marx")
axes[1, 0].set_xlabel("Période t")
axes[1, 0].set_ylabel("Part de marché réalisée")
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# 4. Intensité capitalistique moyenne kappa (Régime biaisé à la Marx)
axes[1, 1].plot(df_marx["t"], df_marx["kappa_shaikh"], label="kappa moyen (Shaikh)", color="tab:red", linewidth=2)
axes[1, 1].plot(df_marx["t"], df_marx["kappa_okishio"], label="kappa moyen (Okishio)", color="tab:gray", linestyle="--")
axes[1, 1].set_title("Mécanisation (kappa moyen) sous innovation biaisée à la Marx")
axes[1, 1].set_xlabel("Période t")
axes[1, 1].set_ylabel("Coût unitaire en capital (kappa)")
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def _best_practice_technique_fixed(self, rule: str, p_ref: float, firms: list) -> tuple:
    """Best-practice propre à chaque règle : coût uc minimal pour Shaikh, taux r maximal pour Okishio."""
    candidates = []
    for f in firms:
        if f.rule == rule:
            candidates.append((f.uc_b, f.kappa_b))
            if f.uc_a is not None:
                candidates.append((f.uc_a, f.kappa_a))
    if not candidates:
        for f in firms:
            candidates.append((f.uc_b, f.kappa_b))
            
    if rule == "shaikh":
        # Shaikh sélectionne la frontière de coût unitaire le plus bas
        return min(candidates, key=lambda tech: tech[0])
    else:
        # Okishio sélectionne le taux de profit instantané le plus élevé
        return max(candidates, key=lambda tech: (p_ref - tech[0]) / tech[1])


def innovate_and_invest_fixed(self, current_t: int, forced_draw: tuple = None):
    """Étape 5 corrigée : pas de gel infini de la R&D et diffusion effective de (uc_a, kappa_a)."""
    # 1. Vérification de la consolidation pour Shaikh
    can_search = True
    if self.rule == "shaikh" and self.uc_a is not None:
        if self._frontier_capacity_share() >= self.model.gamma_thresh:
            self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
            self.uc_a, self.kappa_a = None, None
        else:
            can_search = False

    # 2. Tirage et adoption d'une innovation
    if can_search:
        d_uc, d_kappa = forced_draw if forced_draw is not None else self.model.draw_innovation()
        uc_plus = self.uc_b / (1.0 + d_uc)
        kappa_plus = self.kappa_b / (1.0 + d_kappa)

        if self.rule == "okishio":
            r_b = (self.price - self.uc_b) / self.kappa_b
            r_plus = (self.price - uc_plus) / kappa_plus
            if r_plus > r_b:
                self.uc_b, self.kappa_b = uc_plus, kappa_plus
                self.uc_a, self.kappa_a = None, None
        elif self.rule == "shaikh":
            if uc_plus < self.uc_b:
                self.uc_a, self.kappa_a = uc_plus, kappa_plus

    # 3. Investissement du budget interne B_i(t)
    if self.budget <= 1e-12:
        return

    new_birth = current_t + 1

    if self.rule == "okishio":
        dk_b = self.budget / self.kappa_b
        if dk_b > 0:
            self.vintages.append(Vintage(self.uc_b, self.kappa_b, dk_b, dk_b, new_birth))
            self.budget = 0.0

    elif self.rule == "shaikh":
        if self.uc_a is None:
            dk_b = self.budget / self.kappa_b
            if dk_b > 0:
                self.vintages.append(Vintage(self.uc_b, self.kappa_b, dk_b, dk_b, new_birth))
                self.budget = 0.0
        else:
            # La firme Shaikh oriente son accumulation vers sa technique frontière (uc_a, kappa_a)
            dk_a = self.budget / self.kappa_a
            if dk_a > 0:
                self.vintages.append(Vintage(self.uc_a, self.kappa_a, dk_a, dk_a, new_birth))
                self.budget = 0.0

            # Vérification immédiate si le seuil de consolidation gamma_thresh est atteint
            self.k_surv = self.total_capacity
            if self._frontier_capacity_share() >= self.model.gamma_thresh:
                self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
                self.uc_a, self.kappa_a = None, None


# On attache les deux méthodes corrigées aux classes existantes
FirmAgent.innovate_and_invest = innovate_and_invest_fixed
RealCompetitionModel._best_practice_technique = _best_practice_technique_fixed

# --- Relance de la simulation comparative sur 150 périodes ---
df_neutral = run_simulation(n_steps=150, innovation_regime="neutral", seed=42)
df_marx = run_simulation(n_steps=150, innovation_regime="marx", seed=42)

fig, axes = plt.subplots(2, 2, figsize=(14, 9), sharex=True)

axes[0, 0].plot(df_neutral["t"], df_neutral["p_leader"], label="Leader (Neutre)", color="tab:blue")
axes[0, 0].plot(df_marx["t"], df_marx["p_leader"], label="Leader (Marx-biaisé)", color="tab:red")
axes[0, 0].set_title("Évolution du prix leader p_leader(t)")
axes[0, 0].set_ylabel("Prix")
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

axes[0, 1].plot(df_neutral["t"], df_neutral["r_branch"], label="r_branch (Neutre)", color="tab:blue")
axes[0, 1].plot(df_marx["t"], df_marx["r_branch"], label="r_branch (Marx-biaisé)", color="tab:red")
axes[0, 1].axhline(0.12, color="black", linestyle="--", label="Taux général r* = 12%")
axes[0, 1].set_title("Gravitation du taux de profit de la branche")
axes[0, 1].set_ylabel("Taux de profit")
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

axes[1, 0].plot(df_marx["t"], df_marx["share_shaikh"], label="Part de marché Shaikh", color="tab:red", linewidth=2)
axes[1, 0].plot(df_marx["t"], df_marx["share_okishio"], label="Part de marché Okishio", color="tab:gray", linestyle="--")
axes[1, 0].set_title("Sélection concurrentielle sous innovation biaisée à la Marx")
axes[1, 0].set_xlabel("Période t")
axes[1, 0].set_ylabel("Part de marché réalisée")
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

axes[1, 1].plot(df_marx["t"], df_marx["kappa_shaikh"], label="kappa moyen (Shaikh)", color="tab:red", linewidth=2)
axes[1, 1].plot(df_marx["t"], df_marx["kappa_okishio"], label="kappa moyen (Okishio)", color="tab:gray", linestyle="--")
axes[1, 1].set_title("Mécanisation (kappa moyen) sous innovation biaisée à la Marx")
axes[1, 1].set_xlabel("Période t")
axes[1, 1].set_ylabel("Coût unitaire en capital (kappa)")
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def run_detailed_trajectories(n_steps: int = 100, seed: int = 42) -> pd.DataFrame:
    """Enregistre les trajectoires individuelles de chaque firme (prix, uc, kappa, capacité) en régime Marx."""
    model = RealCompetitionModel(
        D=85.0, u_n=0.85, T=20, gamma=0.03, eps_0=0.005,
        mu_z=0.01, sigma_z=0.005, lam=0.8, R=0.05,
        innovation_regime="marx", marx_a=0.08, marx_theta=-np.pi / 4,
        gamma_thresh=0.75, r_star=0.12, psi=5.0,
        f_0=0.02, K_0=10.0, B_0=100.0, s_min=0.01, tau_grace=20,
        seed=seed
    )

    uc_0, kappa_0 = 0.50, 0.1
    p_0 = 5
    k_init_firm = (85.0 / 0.85) / 10

    for idx in range(10):
        rule = "shaikh" if idx % 2 == 0 else "okishio"
        birth_0 = -(idx % 10)
        v0 = Vintage(uc=uc_0, kappa=kappa_0, k_init=k_init_firm, k=k_init_firm, birth_period=birth_0)
        FirmAgent(
            model=model, rule=rule, initial_vintages=[v0],
            uc_b=uc_0, kappa_b=kappa_0, price=p_0,
            target_share=0.10, budget=15.0, u_init=0.85, entry_period=0
        )

    records = []
    for t in range(1, n_steps + 1):
        model.step()
        for f in model.agents:
            records.append({
                "t": t,
                "firm_id": f.unique_id,
                "rule": f.rule,
                "price": f.price,
                "mean_uc": f.mean_uc,
                "mean_kappa": f.mean_kappa,
                "capacity": f.total_capacity,
                "sales": f.sales,
                "u": f.u,
                "entry_period": f.entry_period
            })

    return pd.DataFrame(records)


df_indiv = run_detailed_trajectories(n_steps=90, seed=42)

fig, axes = plt.subplots(1, 3, figsize=(17, 5.5), sharex=True)

for fid, grp in df_indiv.groupby("firm_id"):
    rule = grp["rule"].iloc[0]
    color = "tab:red" if rule == "shaikh" else "dimgray"
    alpha = 0.75 if grp["entry_period"].iloc[0] == 0 else 0.35
    lw = 1.5 if grp["entry_period"].iloc[0] == 0 else 0.9

    # 1. Prix individuels p_i(t)
    axes[0].plot(grp["t"], grp["price"], color=color, alpha=alpha, linewidth=lw)
    # 2. Coûts unitaires moyens uc_bar_i(t)
    axes[1].plot(grp["t"], grp["mean_uc"], color=color, alpha=alpha, linewidth=lw)
    # 3. Capacités productives K_i(t)
    axes[2].plot(grp["t"], grp["capacity"], color=color, alpha=alpha, linewidth=lw)

# Légende manuelle propre
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color="tab:red", lw=2, label="Firmes Shaikh (trait fin = entrants)"),
    Line2D([0], [0], color="dimgray", lw=2, label="Firmes Okishio (trait fin = entrants)")
]

axes[0].set_title("Trajectoires des prix individuels p_i(t)")
axes[0].set_ylabel("Prix")
axes[0].legend(handles=legend_elements)
axes[0].grid(alpha=0.3)

axes[1].set_title("Coûts unitaires moyens uc_bar_i(t) (Planchers)")
axes[1].set_xlabel("Période t")
axes[1].set_ylabel("Coût unitaire (uc)")
axes[1].grid(alpha=0.3)

axes[2].set_title("Capacités physiques installées K_i(t)")
axes[2].set_xlabel("Période t")
axes[2].set_ylabel("Capacité K")
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def set_price_leader_margin(self, p_leader: float, is_shortage: bool):
    """Étape 1 : epsilon_i(t) dépend de la sous-utilisation ET de la marge relative au prix leader."""
    p_floor = self.mean_uc
    if not is_shortage:
        cap_pressure = self.model.gamma * (self.model.u_n - self.u)
        margin_adv = max(0.0, 1.2*(p_leader - self.mean_uc) / p_leader) if p_leader > 0 else 0.0
        cost_pressure = self.model.delta_cost * margin_adv
        
        eps_i = max(0.0, cap_pressure + cost_pressure + self.model.eps_0)
        self.price = max(p_floor, p_leader * (1.0 - eps_i))
    else:
        z_i = max(0.0, self.model.rng.normal(self.model.mu_z, self.model.sigma_z))
        self.price = max(p_floor, self.price, p_leader * (1.0 + z_i))


def step_1_pricing_leader_margin(self):
    """Applique la fixation de prix avec marge relative au leader."""
    firms = list(self.agents)
    if not firms:
        return
    S_t = sum(f.total_capacity for f in firms)
    p_leader = min(f.price for f in firms)
    is_shortage = S_t < self.D
    for f in firms:
        f.set_price(p_leader=p_leader, is_shortage=is_shortage)


# Mise à jour propre des deux méthodes
RealCompetitionModel.delta_cost = 0.08
FirmAgent.set_price = set_price_leader_margin
FirmAgent.set_price_leader_margin = set_price_leader_margin
RealCompetitionModel.step_1_pricing = step_1_pricing_leader_margin

# --- Lancement de la simulation en régime biaisé à la Marx ---
df_marx_leader = run_simulation(n_steps=150, innovation_regime="marx", D=75.0, seed=42)

fig, axes = plt.subplots(2, 2, figsize=(14, 8.5), sharex=True)

axes[0, 0].plot(df_marx_leader["t"], df_marx_leader["share_shaikh"], label="Part de marché Shaikh", color="tab:red", linewidth=2)
axes[0, 0].plot(df_marx_leader["t"], df_marx_leader["share_okishio"], label="Part de marché Okishio", color="dimgray", linestyle="--", linewidth=2)
axes[0, 0].set_title("Parts de marché (epsilon fonction de p_leader - uc_i)")
axes[0, 0].set_ylabel("Part de marché réalisée")
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

axes[0, 1].plot(df_marx_leader["t"], df_marx_leader["p_leader"], label="Prix leader p_leader(t)", color="tab:red", linewidth=2)
axes[0, 1].plot(df_marx_leader["t"], df_marx_leader["p_avg"], label="Prix moyen p_avg(t)", color="black", linestyle=":")
axes[0, 1].set_title("Trajectoire des prix")
axes[0, 1].set_ylabel("Prix")
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

axes[1, 0].plot(df_marx_leader["t"], df_marx_leader["r_branch"], label="r_branch", color="tab:red", linewidth=2)
axes[1, 0].axhline(0.12, color="black", linestyle="--", label="Taux général r* = 12%")
axes[1, 0].set_title("Gravitation du taux de profit de la branche")
axes[1, 0].set_xlabel("Période t")
axes[1, 0].set_ylabel("Taux de profit")
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

axes[1, 1].plot(df_marx_leader["t"], df_marx_leader["kappa_shaikh"], label="kappa moyen (Shaikh)", color="tab:red", linewidth=2)
axes[1, 1].plot(df_marx_leader["t"], df_marx_leader["kappa_okishio"], label="kappa moyen (Okishio)", color="dimgray", linestyle="--", linewidth=2)
axes[1, 1].set_title("Mécanisation (kappa moyen)")
axes[1, 1].set_xlabel("Période t")
axes[1, 1].set_ylabel("Coût unitaire en capital (kappa)")
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def run_epsilon_diagnostics(n_steps: int = 150, D: float = 75.0, seed: int = 42) -> pd.DataFrame:
    """Enregistre l'évolution de epsilon_i(t) et de l'utilisation u_i(t) pour Shaikh vs Okishio."""
    model = RealCompetitionModel(
        D=D, u_n=0.85, T=10, gamma=0.03, eps_0=0.005,
        mu_z=0.01, sigma_z=0.005, lam=0.8, R=0.05,
        innovation_regime="marx", marx_a=0.08, marx_theta=-np.pi / 4,
        gamma_thresh=0.75, r_star=0.12, psi=5.0,
        f_0=0.02, K_0=10.0, B_0=15.0, s_min=0.01, tau_grace=20,
        seed=seed
    )

    uc_0, kappa_0 = 0.50, 1.0
    p_0 = uc_0 + 0.12 * kappa_0
    k_init_firm = (D / 0.85) / 10

    for idx in range(10):
        rule = "shaikh" if idx % 2 == 0 else "okishio"
        birth_0 = -(idx % 10)
        v0 = Vintage(uc=uc_0, kappa=kappa_0, k_init=k_init_firm, k=k_init_firm, birth_period=birth_0)
        FirmAgent(
            model=model, rule=rule, initial_vintages=[v0],
            uc_b=uc_0, kappa_b=kappa_0, price=p_0,
            target_share=0.10, budget=15.0, u_init=0.85, entry_period=0
        )

    history = []
    for t in range(1, n_steps + 1):
        # 1. On calcule epsilon_i(t) juste avant step() (car il dépend de u_i(t-1))
        firms_before = list(model.agents)
        S_t = sum(f.total_capacity for f in firms_before)
        is_shortage = S_t < model.D

        eps_by_id = {}
        for f in firms_before:
            if not is_shortage:
                eps_by_id[f.unique_id] = max(0.0, model.gamma * (model.u_n - f.u) + model.eps_0)
            else:
                eps_by_id[f.unique_id] = 0.0  # En pénurie, pas de sous-cotation (regime z_i)

        # 2. Exécution de la période t
        model.step()

        # 3. Collecte par groupe (on garde les firmes présentes au début de la période)
        shaikh_eps = [eps_by_id[f.unique_id] for f in firms_before if f.rule == "shaikh"]
        okishio_eps = [eps_by_id[f.unique_id] for f in firms_before if f.rule == "okishio"]

        firms_after = list(model.agents)
        shaikh_after = [f for f in firms_after if f.rule == "shaikh"]
        okishio_after = [f for f in firms_after if f.rule == "okishio"]

        history.append({
            "t": t,
            "is_shortage": int(is_shortage),
            "eps_shaikh_mean": np.mean(shaikh_eps) if shaikh_eps else np.nan,
            "eps_shaikh_max": np.max(shaikh_eps) if shaikh_eps else np.nan,
            "eps_okishio_mean": np.mean(okishio_eps) if okishio_eps else np.nan,
            "eps_okishio_max": np.max(okishio_eps) if okishio_eps else np.nan,
            "u_shaikh_mean": np.mean([f.u for f in shaikh_after]) if shaikh_after else np.nan,
            "u_okishio_mean": np.mean([f.u for f in okishio_after]) if okishio_after else np.nan,
            "share_shaikh": sum(f.realized_share for f in shaikh_after),
            "share_okishio": sum(f.realized_share for f in okishio_after),
        })

    return pd.DataFrame(history)


df_eps = run_epsilon_diagnostics(n_steps=150, D=75.0, seed=42)

fig, axes = plt.subplots(3, 1, figsize=(13, 11), sharex=True)

# --- Panneau 1 : Agressivité tarifaire epsilon_i(t) (Moyenne et Max) ---
axes[0].plot(df_eps["t"], df_eps["eps_shaikh_mean"], label="epsilon moyen (Shaikh)", color="tab:red", linewidth=2)
axes[0].plot(df_eps["t"], df_eps["eps_shaikh_max"], label="epsilon max (Shaikh)", color="tab:red", linestyle=":", alpha=0.8)
axes[0].plot(df_eps["t"], df_eps["eps_okishio_mean"], label="epsilon moyen (Okishio)", color="dimgray", linewidth=2)
axes[0].plot(df_eps["t"], df_eps["eps_okishio_max"], label="epsilon max (Okishio)", color="dimgray", linestyle=":", alpha=0.8)
axes[0].axhline(0.005, color="black", linestyle="--", alpha=0.4, label="epsilon_0 = 0.005")
axes[0].set_title("Agressivité de baisse des prix epsilon_i(t) (0 = régime de pénurie S(t) < D(t))")
axes[0].set_ylabel("epsilon_i(t)")
axes[0].legend(loc="upper right", ncol=2)
axes[0].grid(alpha=0.3)

# --- Panneau 2 : Taux d'utilisation des capacités u_i(t) ---
axes[1].plot(df_eps["t"], df_eps["u_shaikh_mean"], label="Utilisation moyenne u(t) (Shaikh)", color="tab:red", linewidth=2)
axes[1].plot(df_eps["t"], df_eps["u_okishio_mean"], label="Utilisation moyenne u(t) (Okishio)", color="dimgray", linewidth=2)
axes[1].axhline(0.85, color="black", linestyle="--", alpha=0.5, label="Utilisation normale u_n = 0.85")
axes[1].set_title("Taux d'utilisation des capacités u_i(t)")
axes[1].set_ylabel("Utilisation u(t)")
axes[1].legend(loc="lower right")
axes[1].grid(alpha=0.3)

# --- Panneau 3 : Parts de marché réalisées ---
axes[2].plot(df_eps["t"], df_eps["share_shaikh"], label="Part de marché Shaikh", color="tab:red", linewidth=2)
axes[2].plot(df_eps["t"], df_eps["share_okishio"], label="Part de marché Okishio", color="dimgray", linestyle="--", linewidth=2)
axes[2].set_title("Parts de marché réalisées")
axes[2].set_xlabel("Période t")
axes[2].set_ylabel("Part de marché")
axes[2].legend(loc="center right")
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def innovate_and_invest_symmetric(self, current_t: int, forced_draw: tuple = None):
    """Étape 5 avec consolidation symétrique (gamma_thresh) pour Shaikh ET Okishio."""
    # 1. Discipline de consolidation identique pour les deux règles
    can_search = True
    if self.uc_a is not None:
        if self._frontier_capacity_share() >= self.model.gamma_thresh:
            # La technique frontière est consolidée -> elle devient la nouvelle référence
            self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
            self.uc_a, self.kappa_a = None, None
        else:
            # Diffusion interne en cours : R&D suspendue pour Shaikh comme pour Okishio
            can_search = False

    # 2. Tirage et critère d'adoption propre à chaque règle
    if can_search:
        d_uc, d_kappa = forced_draw if forced_draw is not None else self.model.draw_innovation()
        uc_plus = self.uc_b / (1.0 + d_uc)
        kappa_plus = self.kappa_b / (1.0 + d_kappa)

        if self.rule == "okishio":
            r_b = (self.price - self.uc_b) / self.kappa_b
            r_plus = (self.price - uc_plus) / kappa_plus
            if r_plus > r_b:
                self.uc_a, self.kappa_a = uc_plus, kappa_plus
        elif self.rule == "shaikh":
            if uc_plus < self.uc_b:
                self.uc_a, self.kappa_a = uc_plus, kappa_plus

    # 3. Investissement du budget interne B_i(t) dans la technique active la plus avancée
    if self.budget <= 1e-12:
        return

    new_birth = current_t + 1
    target_uc = self.uc_a if self.uc_a is not None else self.uc_b
    target_kappa = self.kappa_a if self.kappa_a is not None else self.kappa_b

    dk = self.budget / target_kappa
    if dk > 0:
        self.vintages.append(Vintage(target_uc, target_kappa, dk, dk, new_birth))
        self.budget = 0.0

    # 4. Vérification immédiate si l'investissement de la période permet d'atteindre gamma_thresh
    if self.uc_a is not None:
        self.k_surv = self.total_capacity
        if self._frontier_capacity_share() >= self.model.gamma_thresh:
            self.uc_b, self.kappa_b = self.uc_a, self.kappa_a
            self.uc_a, self.kappa_a = None, None


# Application de la règle symétrique et relance de la simulation
FirmAgent.innovate_and_invest = innovate_and_invest_symmetric

df_sym = run_simulation(n_steps=150, innovation_regime="marx", D=75.0, seed=42)

fig, axes = plt.subplots(2, 2, figsize=(14, 8.5), sharex=True)

axes[0, 0].plot(df_sym["t"], df_sym["share_shaikh"], label="Part de marché Shaikh", color="tab:red", linewidth=2)
axes[0, 0].plot(df_sym["t"], df_sym["share_okishio"], label="Part de marché Okishio", color="dimgray", linestyle="--", linewidth=2)
axes[0, 0].set_title("Parts de marché (Consolidation symétrique + marge leader)")
axes[0, 0].set_ylabel("Part de marché réalisée")
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

axes[0, 1].plot(df_sym["t"], df_sym["p_leader"], label="Prix leader p_leader(t)", color="tab:red", linewidth=2)
axes[0, 1].plot(df_sym["t"], df_sym["p_avg"], label="Prix moyen p_avg(t)", color="black", linestyle=":")
axes[0, 1].set_title("Trajectoire des prix")
axes[0, 1].set_ylabel("Prix")
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

axes[1, 0].plot(df_sym["t"], df_sym["r_branch"], label="r_branch", color="tab:red", linewidth=2)
axes[1, 0].axhline(0.12, color="black", linestyle="--", label="Taux général r* = 12%")
axes[1, 0].set_title("Gravitation du taux de profit de la branche")
axes[1, 0].set_xlabel("Période t")
axes[1, 0].set_ylabel("Taux de profit")
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

axes[1, 1].plot(df_sym["t"], df_sym["kappa_shaikh"], label="kappa moyen (Shaikh)", color="tab:red", linewidth=2)
axes[1, 1].plot(df_sym["t"], df_sym["kappa_okishio"], label="kappa moyen (Okishio)", color="dimgray", linestyle="--", linewidth=2)
axes[1, 1].set_title("Mécanisation (kappa moyen)")
axes[1, 1].set_xlabel("Période t")
axes[1, 1].set_ylabel("Coût unitaire en capital (kappa)")
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
